```python
import numpy as np
import pandas as pd

# 参数设置
n_samples = 1000                 # 生成的随机欧拉角数量
seed = 42                        # 随机数种子，保证可重复
mode = "uniform_SO3"              # "uniform_SO3" 或 "uniform_angles"
out_csv = "euler_angles.csv"      # 输出 CSV 文件名

# 如果选择 uniform_angles，这里定义范围
phi1_range = (0.0, 360.0)
Phi_range  = (0.0, 180.0)
phi2_range = (0.0, 360.0)

# 初始化随机数
rng = np.random.default_rng(seed)

def sample_uniform_SO3(n, rng):
    """在 SO(3) 上均匀采样 (Bunge convention, degrees)."""
    phi1 = rng.uniform(0, 360, n)
    phi2 = rng.uniform(0, 360, n)
    u = rng.uniform(-1, 1, n)
    Phi = np.degrees(np.arccos(u))
    return phi1, Phi, phi2

def sample_uniform_angles(n, rng, r1, rPhi, r2):
    """简单的角度范围均匀采样."""
    phi1 = rng.uniform(r1[0], r1[1], n)
    Phi  = rng.uniform(rPhi[0], rPhi[1], n)
    phi2 = rng.uniform(r2[0], r2[1], n)
    return phi1, Phi, phi2

# 采样
if mode == "uniform_SO3":
    phi1, Phi, phi2 = sample_uniform_SO3(n_samples, rng)
else:
    phi1, Phi, phi2 = sample_uniform_angles(n_samples, rng, phi1_range, Phi_range, phi2_range)

# 保存到 CSV
df = pd.DataFrame({"phi1_deg": phi1, "Phi_deg": Phi, "phi2_deg": phi2})
df.to_csv(out_csv, index=False)

print(f"生成完毕，已保存到 {out_csv}")
print(df.head())
```
